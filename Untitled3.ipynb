```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as  plt
```

```python
df =pd.read_csv( r"C:\Users\admin\Downloads\Movie_classification.csv ", header= 0)
```

```python
df.head()
```

```python
 Getting null values
```

```python
df.info()
```

```python
 Filling the null values with the mean
```

```python
df['Time_taken'] = df['Time_taken'].fillna(df['Time_taken'].mean())
```

```python
df.info()
```

```python
 The missing value in Time_taken column is filled, next is to convert categorical values into dummy values
as done in data preprocessing
```

```python
df=pd.get_dummies(df,columns=['3D_available', 'Genre'], drop_first=True)
```

```python
df.head()
```

```python
 Now lets split the data into X-Y
```

```python
x = df.loc[:, df.columns != "Start_Tech_Oscar"]
```

```python
x.head()
```

```python
y =df["Start_Tech_Oscar"]
```

```python
y.head()
```

```python
y.shape
```

```python
 TRAIN TEST SPLIT
```

```python
 from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split (x,y, test_size=0.2, random_state=0)
```

```python
x_train.head()
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
   TRAINING A CLASSIFIER TREE
```

```python
from sklearn import tree
```

```python
Oto = tree.DecisionTreeClassifier(max_depth =3)
```

```python
Oto.fit(x_train,y_train)
```

```python
from sklearn import set_config

set_config(display='text')
```

```python
y_train_pred = Oto.predict(x_train)
y_test_pred = Oto.predict(x_test)
    
```

```python
y_train_pred
```

```python
y_test_pred
```

```python
Oto
```

```python
 EVALUATING MODEL PERFORMANCE
```

```python
 from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
confusion_matrix(y_train, y_train_pred)
```

```python
confusion_matrix(y_test, y_test_pred)
```

```python
accuracy_score(y_test, y_test_pred)
```
